```python
import matplotlib.pyplot as plt
import seaborn as sns

print("Visualization libraries imported successfully!")
```

```python
# Calculate sales by category
category_sales = df.groupby('Category')['Sales'].sum()

# Create bar chart
plt.figure(figsize=(8,5))
category_sales.plot(kind='bar')

plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
category_sales.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(7,7)
)

plt.title('Sales Distribution by Category')
plt.ylabel('')
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Sales'], bins=20)

plt.title('Sales Distribution')
plt.xlabel('Sales Amount')
plt.ylabel('Frequency')

plt.show()
```

```python

category_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(7, 7))
plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Sales Distribution by Category')
plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.hist(df['Sales'], bins=20, edgecolor='black')

plt.title('Sales Distribution')
plt.xlabel('Sales Amount')
plt.ylabel('Frequency')

plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.boxplot(x=df['Sales'])

plt.title('Sales Distribution and Outliers')
plt.xlabel('Sales Amount')

plt.show()
```

```python

if 'Region' in df.columns:
    region_sales = df.groupby('Region')['Sales'].sum()

    plt.figure(figsize=(8, 5))
    region_sales.plot(kind='bar', edgecolor='black')

    plt.title('Total Sales by Region')
    plt.xlabel('Region')
    plt.ylabel('Total Sales')
    plt.xticks(rotation=0)
    plt.show()
else:
    print("Region column is not available in this dataset.")
```

```python
plt.figure(figsize=(8, 6))

numeric_data = df.select_dtypes(include='number')

correlation = numeric_data.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Sales Dataset Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x='Sales',
    y='Profit',
    hue='Category',
    alpha=0.7
)

plt.title('Sales vs Profit by Category')
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Set dashboard style
sns.set_theme(style="whitegrid")

# Calculate important business values
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()

if 'Order ID' in df.columns:
    total_orders = df['Order ID'].nunique()
else:
    total_orders = len(df)

# Prepare category sales
category_sales = df.groupby('Category')['Sales'].sum()

# Prepare regional sales if available
if 'Region' in df.columns:
    region_sales = df.groupby('Region')['Sales'].sum()

# Create dashboard layout
fig = plt.figure(figsize=(16, 10))
fig.suptitle(
    'SALES PERFORMANCE ANALYSIS DASHBOARD',
    fontsize=20,
    fontweight='bold'
)

gs = fig.add_gridspec(3, 2, height_ratios=[0.7, 1.5, 1.5])

# KPI 1: Total Sales
ax1 = fig.add_subplot(gs[0, 0])
ax1.text(0.5, 0.6, f'${total_sales:,.2f}',
         ha='center', fontsize=22, fontweight='bold')
ax1.text(0.5, 0.2, 'TOTAL SALES',
         ha='center', fontsize=13)
ax1.axis('off')

# KPI 2: Total Profit
ax2 = fig.add_subplot(gs[0, 1])
ax2.text(0.5, 0.6, f'${total_profit:,.2f}',
         ha='center', fontsize=22, fontweight='bold')
ax2.text(0.5, 0.2, 'TOTAL PROFIT',
         ha='center', fontsize=13)
ax2.axis('off')

# Chart 1: Sales by Category
ax3 = fig.add_subplot(gs[1, 0])
category_sales.plot(kind='bar', ax=ax3)
ax3.set_title('Sales by Product Category')
ax3.set_xlabel('Category')
ax3.set_ylabel('Sales Amount')
ax3.tick_params(axis='x', rotation=0)

# Chart 2: Sales by Region
ax4 = fig.add_subplot(gs[1, 1])

if 'Region' in df.columns:
    region_sales.plot(kind='bar', ax=ax4)
    ax4.set_title('Sales by Region')
    ax4.set_xlabel('Region')
    ax4.set_ylabel('Sales Amount')
    ax4.tick_params(axis='x', rotation=0)
else:
    ax4.text(0.5, 0.5, 'Region data unavailable',
             ha='center', va='center')
    ax4.set_title('Sales by Region')

# Chart 3: Sales Distribution
ax5 = fig.add_subplot(gs[2, 0])
ax5.hist(df['Sales'], bins=20, edgecolor='black')
ax5.set_title('Sales Distribution')
ax5.set_xlabel('Sales Amount')
ax5.set_ylabel('Frequency')

# Chart 4: Profit by Category
ax6 = fig.add_subplot(gs[2, 1])
profit_category = df.groupby('Category')['Profit'].sum()
profit_category.plot(kind='bar', ax=ax6)
ax6.set_title('Profit by Product Category')
ax6.set_xlabel('Category')
ax6.set_ylabel('Profit')
ax6.tick_params(axis='x', rotation=0)

plt.tight_layout(rect=[0, 0, 1, 0.95])

# Save dashboard image
plt.savefig('CodeAlpha_Sales_Dashboard.png',
            dpi=300, bbox_inches='tight')

plt.show()

print("Dashboard created successfully!")
print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders / Records:", total_orders)
```

```python

# 1. Total Sales and Profit
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))

# 2. Highest Sales Category
category_sales = df.groupby('Category')['Sales'].sum()

print("\nSales by Category:")
print(category_sales)

print("\nHighest Sales Category:")
print(category_sales.idxmax())

# 3. Highest Profit Category
category_profit = df.groupby('Category')['Profit'].sum()

print("\nProfit by Category:")
print(category_profit)

print("\nHighest Profit Category:")
print(category_profit.idxmax())

# 4. Highest Sales Region
if 'Region' in df.columns:
    region_sales = df.groupby('Region')['Sales'].sum()

    print("\nSales by Region:")
    print(region_sales)

    print("\nHighest Sales Region:")
    print(region_sales.idxmax())
else:
    print("\nRegion column is not available.")

# 5. Average Sales
print("\nAverage Sales:", round(df['Sales'].mean(), 2))

# 6. Highest and Lowest Sales
print("Highest Sale:", df['Sales'].max())
print("Lowest Sale:", df['Sales'].min())
```

```python

# Calculate sales quartiles
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)

IQR = Q3 - Q1

# Identify potential outliers
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df['Sales'] < lower_limit) |
    (df['Sales'] > upper_limit)
]

print("Potential Sales Outliers:", len(outliers))
print("\nOutlier Records:")
display(outliers.head(10))
```
